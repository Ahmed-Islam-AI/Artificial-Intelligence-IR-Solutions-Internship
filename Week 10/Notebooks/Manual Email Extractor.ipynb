```python
# Filter rows where email is missing (NaN or empty string)
missing_email_df = df[df['Emails'].isna() | (df['Emails'].astype(str).str.strip() == '')]

# Extract only the domain column
domains_missing_email = missing_email_df[['Domain']]

# Save to new Excel/CSV file
output_file = "domains_missing_email.xlsx"
domains_missing_email.to_excel(output_file, index=False)

print(f"✅ Extracted {len(domains_missing_email)} domains with missing emails and saved to {output_file}")
```

```python
import pandas as pd
import requests
from bs4 import BeautifulSoup
import re
import time

# Function to extract emails from a given URL
def extract_emails_from_url(url):
    emails = set()
    try:
        headers = {"User-Agent": "Mozilla/5.0"}
        response = requests.get(url, headers=headers, timeout=10)
        if response.status_code == 200:
            soup = BeautifulSoup(response.text, "html.parser")
            text = soup.get_text(" ", strip=True)

            # Regex for email
            found_emails = re.findall(r"[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}", text)
            emails.update(found_emails)

            # Check <a href="mailto:...">
            for mailto in soup.select('a[href^=mailto]'):
                emails.add(mailto['href'].replace("mailto:", "").strip())
    except Exception as e:
        print(f"❌ Error fetching {url}: {e}")
    return list(emails)

# Function to check homepage + common pages
def find_emails(domain):
    urls_to_check = [
        f"https://{domain}",
        f"http://{domain}",
        f"https://{domain}/contact",
        f"https://{domain}/about",
        f"https://{domain}/support",
        f"http://{domain}/contact-us",
        f"http://{domain}/about-us",
    ]

    for url in urls_to_check:
        emails = extract_emails_from_url(url)
        if emails:
            return emails[0]  # return the first email found
    return None

# === Main ===
input_file = "/content/domains_missing_email.xlsx"   # Your input file
output_file = "domains_with_emails.csv"

# Load domains
df = pd.read_excel(input_file)

# Extract emails for each domain
extracted_emails = []
for domain in df["Domain"]:
    email = find_emails(domain)
    extracted_emails.append(email)
    print(f"🌍 Domain: {domain}  |  📧 Email: {email}")  # Print progress
    time.sleep(2)  # polite delay to avoid blocking (adjust if needed)

df["extracted_email"] = extracted_emails

# Save to CSV
df.to_csv(output_file, index=False)

print(f"\n✅ Emails extracted and saved to {output_file}")
```
