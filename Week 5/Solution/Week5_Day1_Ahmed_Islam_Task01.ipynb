# Artificial Neural Networks From Scratch using Numpy and Python

```python
# necessary library
import numpy as np
```

```python
# function for the weight and biases
def initialize_parameters(input_size, hidden_size, output_size, seed=42):
    np.random.seed(seed)
    W1 = np.random.randn(hidden_size, input_size) * 0.01
    b1 = np.zeros((hidden_size, 1))
    W2 = np.random.randn(output_size, hidden_size) * 0.01
    b2 = np.zeros((output_size, 1))
    return W1, b1, W2, b2
```

# Activation Function   


```python
# for output layer
def sigmoid(x):
    return 1 / (1 + np.exp(-x))  # formula of the sigmoid function for binary classification output will be between 0 or 1

# for hidden layer
def relu(x):
    return np.maximum(0, x) # relu basically check if the value is less than zero it assign a 0 value otherwise its orignal value pass

# for backpropagation
def relu_derivative(x):
    return (x > 0).astype(float) #Derivative of ReLU. Needed for backprop. Returns 1 where x > 0, else 0
```

# Forward Propagation

```python
def forward_propagation(X, W1, b1, W2, b2):
    Z1 = np.dot(W1, X) + b1    # dot product of x and w1 and than adding the bias
    A1 = relu(Z1)      # adding the non linearity
    Z2 = np.dot(W2, A1) + b2
    A2 = sigmoid(Z2)

    # cache for the back propagation
    cache = {
        "Z1": Z1,
        "A1": A1,
        "Z2": Z2,
        "A2": A2
        }
    return A2, cache
```

# Compute the cost function


```python
def compute_cost(A2, Y):   # A2 is predicted output and Y is actual label
     m = Y.shape[1]        # m is basically number of examples

     epsilon = 1e-8  # to prevent log(0)
     epsilon = 1e-8  # to prevent log(0)
     cost = -np.sum(Y * np.log(A2) + (1 - Y) * np.log(1 - A2)) / m   # formula for calculating the cost or error
     return float(cost)
```

# Back Propagation

```python
def backward_propagation(X, Y, cache, W2):
    m = X.shape[1]
    A1, A2, Z1 = cache["A1"], cache["A2"], cache["Z1"]  # accessing the values from dectionary

    dZ2 = A2 - Y   # error in predicted and actual label
    dW2 = (1 / m) * np.dot(dZ2, A1.T)    # gradient cost w.r.t w1
    db2 = (1 / m) * np.sum(dZ2, axis=1, keepdims=True)   # gradient cost w.r.t b2

    dA1 = np.dot(W2.T, dZ2)    # propagate error back to hidden layer
    dZ1 = dA1 * relu_derivative(Z1)   #for backpropagation
    dW1 = (1 / m) * np.dot(dZ1, X.T)    # gradient for first layer
    db1 = (1 / m) * np.sum(dZ1, axis=1, keepdims=True)

    return dW1, db1, dW2, db2
```

```python
def update_parameters(W1, b1, W2, b2, dW1, db1, dW2, db2, lr):
    W1 -= lr * dW1
    b1 -= lr * db1
    W2 -= lr * dW2
    b2 -= lr * db2
    return W1, b1, W2, b2
```

# Training the Neural Network

```python
def train_neural_network(X, Y, hidden_size=2, iterations=10000, learning_rate=1.2):
    input_size = X.shape[0]
    output_size = Y.shape[0]
    W1, b1, W2, b2 = initialize_parameters(input_size, hidden_size, output_size)

    for i in range(iterations):
        A2, cache = forward_propagation(X, W1, b1, W2, b2)
        cost = compute_cost(A2, Y)
        dW1, db1, dW2, db2 = backward_propagation(X, Y, cache, W2)
        W1, b1, W2, b2 = update_parameters(W1, b1, W2, b2, dW1, db1, dW2, db2, learning_rate)

        if i % 1000 == 0:
            print(f"Iteration {i}, Cost: {cost:.4f}")

    return {"W1": W1, "b1": b1, "W2": W2, "b2": b2}
```

# Prediction Function

```python
def predict(X, parameters):
    W1, b1, W2, b2 = parameters["W1"], parameters["b1"], parameters["W2"], parameters["b2"]
    A2, _ = forward_propagation(X, W1, b1, W2, b2)   # forward pas with traning weights
    predictions = (A2 > 0.5).astype(int)   # threashold value for the classes prediction
    return predictions
```

# Final Execution - Train and test on AND Gate


```python
X = np.array([[0, 0, 1, 1],
              [0, 1, 0, 1]])    # 4 example, each with 2 features

Y = np.array([0, 0, 0, 1]).reshape(1, 4)    # AND gate output

parameters = train_neural_network(X, Y, hidden_size=2, iterations=5000, learning_rate=1.0)

preds = predict(X, parameters)
print("Predictions:", preds)
```
